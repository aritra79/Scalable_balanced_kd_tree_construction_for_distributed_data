## Type A Transformation 

This function implements the main functional component of type A transformation involving Chebyshev's polynomials to a 1D array, as discussed in _Chakravorty, A., et al. (2021)_.

### Parameters

- **ndarray**  
  `u`: 1D input array $\boldsymbol{u}=\big(u_1, u_2,\dots, u_J\big)'\in\mathbb{R}^J$.

### Returns

- **ndarray**  
  `T`: The transformed 1D array $\mathcal{T}_J^{(A)}(\boldsymbol{u})\in\mathbb{R}^J$, the index $J$ in $\mathcal{T}_J^{(A)}$ means it expects 1D input array of $J$ elements.

### Notes

- The function implements a linear-transformation $\mathcal{T}_J^{(A)}:\mathbb{R}^J\to\mathbb{R}^J$, equivalent to left-multiplication by a $J\times J$ lower-triangular matrix $\mathsf{A}_J^{(A)}$, which can be expressed in terms of the coefficients from Chebyshev's orthogonal polynomials. This is formally verified in _Chakravorty, A., et al. (2021)_.  
- When the input is the odd power basis vector $\boldsymbol{u}\,(t)=\big(t^1,t^3,\dots,t^{2J-1}\big)'$, the output is the odd-degree Chebyshev polynomial basis $\mathcal{T}_J^{(A)}\big(\boldsymbol{u}\,(t)\big)=\big(T_1(t),T_3(t),\dots,T_{2J-1}(t)\big)'$, where $T_{n}(t)$ is the $n$th Chebyshev polynomial of the first kind. Also, for $x\in\mathbb{R}$ and $j\in\mathbb{Z}_{> 0}$: $T_{2j-1}\big(\cos(x)\big)=\cos\big((2j-1)\,x\big)$ and $T_{2j-1}\big(\sin(x)\big)=(-1)^{j-1}\,\sin\big((2j-1)\,x\big)$.
- For the canonical input $\boldsymbol{u}\,(t)=\big(t^1,t^3,\dots,t^{2J-1}\big)'$, observe that $T_0(t)=1$ and $T_1(t)=t$. Consequently, we start the transformation with two $J$-dimensional vectors: $\big(t^1\,T_0(t),t^3\,T_0(t),\dots,t^{2J-1}\,T_0(t)\big)'$ and $\big(t^0\,T_1(t),t^2\,T_1(t),\dots,t^{2J-2}\,T_1(t)\big)'$. The transformation consists of $J$ iterative steps as follows:
    - We start the $j$th iteration with $(J-j+1)$-dimensional vectors: $\big(t^1\,T_{2j-2}(t),t^3\,T_{2j-2}(t),\dots,t^{2J-2j+1}\,T_{2j-2}(t)\big)'$ and $\big(t^0\,T_{2j-1}(t),t^2\,T_{2j-1}(t),\dots,t^{2J-2j}\,T_{2j-1}(t)\big)'$, we add $T_{2j-1}(t)$ as the $j$th element of the final-output vector.
    - Next, use the recurrence relation: $t^{2k-1}\,T_{2j}(t)=2\,t^{2k}\,T_{2j-1}(t)-t^{2k-1}\,T_{2j-2}(t)$, to calculate: $t\,T_{2j}(t),t^3\,T_{2j}(t),\cdots,t^{2J-2j-1}\,T_{2j}(t)$ from $t\,T_{2j-2}(t),t^3\,T_{2j-2}(t),\cdots,t^{2J-2j-1}\,T_{2j-2}(t)$ and $t^2\,T_{2j-1}(t),t^4\,T_{2j-1}(t),\cdots,t^{2J-2j}\,T_{2j-1}(t)$. The following diagram demonstrates this process:
\begin{array}{ccccc}
 & -1\,\times\,t\,T_{2j-2}(t) & -1\,\times\,t^3\,T_{2j-2}(t) & \cdots &  -1\,\times\,t^{2J-2j-1}\,T_{2j-2}(t) \\
 & + & + & & + \\
 & \quad 2\,\times\,t^2\,T_{2j-1}(t) & \quad 2\,\times\,t^4\,T_{2j-1}(t) & \cdots & \quad 2\,\times\,t^{2J-2j}\,T_{2j-1}(t) \\
 & \shortparallel & \shortparallel & & \shortparallel \\
 & t\,T_{2j}(t) & t^3\,T_{2j}(t) & \cdots &  t^{2J-2j-1}\,T_{2j}(t)
\end{array}
    - Finally, we use the recurrence relation: $t^{2k-2}\,T_{2j+1}(t)=2\,t^{2k-1}\,T_{2j}(t)-t^{2k-2}\,T_{2j-1}(t)$, to calculate: $T_{2j+1}(t),t^2\,T_{2j+1}(t),\cdots,t^{2J-2j-2}\,T_{2j+1}(t)$ from $T_{2j-1}(t),t^2\,T_{2j-1}(t),\cdots,t^{2J-2j-2}\,T_{2j-1}(t)$ and $t\,T_{2j}(t),t^3\,T_{2j}(t),\cdots,t^{2J-2j-1}\,T_{2j}(t)$. The following diagram demonstrates this process:
\begin{array}{ccccc}
 & -1\,\times\,T_{2j-1}(t) & -1\,\times\,t^2\,T_{2j-1}(t) & \cdots &  -1\,\times\,t^{2J-2j-2}\,T_{2j-1}(t) \\
 & + & + & & + \\
 & \quad 2\,\times\,t\,T_{2j}(t) & \quad 2\,\times\,t^3\,T_{2j}(t) & \cdots & \quad 2\,\times\,t^{2J-2j-1}\,T_{2j}(t) \\
 & \shortparallel & \shortparallel & & \shortparallel \\
 & T_{2j+1}(t) & t^2\,T_{2j+1}(t) & \cdots &  t^{2J-2j-2}\,T_{2j+1}(t)
\end{array}
- For a general input $\boldsymbol{u}=\big(u_1,u_2,\dots,u_J\big)'$, we set $\omega_{0,\,2k-1}=\omega_{1,\,2k-2}=u_k$ for $k=1,2,\dots,J$, and start the transformation with $\big(\omega_{0,\,1},\omega_{0,\,3},\dots,\omega_{0,\,2J-1}\big)'$ and $\big(\omega_{1,\,0},\omega_{1,\,2},\dots,\omega_{1,\,2J-2}\big)'$ and perform $J$ iterative steps, mimicking the logic for canonical input as follows:
    - We start the $j$th iteration with $(J-j+1)$-dimensional vectors: $\big(\omega_{2j-2,\,1},\omega_{2j-2,\,3},\dots,\omega_{2j-2,\,2J-2j+1}\big)'$ and $\big(\omega_{2j-1,\,0},\omega_{2j-1,\,2},\dots,\omega_{2j-1,\,2J-2j}\big)'$, we add $w_j=\omega_{2j-1,\,0}$ as the $j$th element of the final-output vector.
    - Next, use the recurrence relation: $\omega_{2j,\,2k-1} = 2\,\omega_{2j-1,\,2k} - \omega_{2j-2,\,2k-1}$, to calculate $\omega_{2j,\,1},\,\omega_{2j,\,3},\,\cdots,\,\omega_{2j,\,2J-2j-1}$ from $\omega_{2j-2,\,1},\,\omega_{2j-2,\,3},\,\cdots,\,\omega_{2j-2,\,2J-2j-1}$ and $\omega_{2j-1,\,2},\,\omega_{2j-1,\,4},\,\cdots,\,\omega_{2j-1,\,2J-2j}$. The following diagram demonstrates this process:
\begin{array}{ccccc}
 & -1\times\omega_{2j-2,\,1} & -1\times\omega_{2j-2,\,3} & \cdots & -1\times\omega_{2j-2,\,2J-2j-1} \\
 & + & + & & + \\
 & \quad 2\times\omega_{2j-1,\,2} & \quad 2\times\omega_{2j-1,\,4} & \cdots & \quad 2\times\omega_{2j-1,\,2J-2j} \\
 & \shortparallel & \shortparallel & & \shortparallel \\
 & \omega_{2j,\,1} & \omega_{2j,\,3} & \cdots & \omega_{2j,\,2J-2j-1}
\end{array}
    - Finally, we use the recurrence relation: $\omega_{2j+1,\,2k-2} = 2\,\omega_{2j,\,2k-1} - \omega_{2j-1,\,2k-2}$, to calculate $\omega_{2j+1,\,0},\,\omega_{2j+1,\,2},\,\cdots,\,\omega_{2j+1,\,2J-2j-2}$ from $\omega_{2j-1,\,0},\,\omega_{2j-1,\,2},\,\cdots,\,\omega_{2j-1,\,2J-2j-2}$ and $\omega_{2j,\,1},\,\omega_{2j,\,3},\,\cdots,\,\omega_{2j,\,2J-2j-1}$. The following diagram demonstrates this process:\begin{array}{ccccc}
 & -1\times\omega_{2j-1,\,0} & -1\times\omega_{2j-1,\,2} & \cdots & -1\times\omega_{2j-1,\,2J-2j-2} \\
 & + & + & & + \\
 & \quad 2\times\omega_{2j,\,1} & \quad 2\times\omega_{2j,\,3} & \cdots & \quad 2\times\omega_{2j,\,2J-2j-1} \\
 & \shortparallel & \shortparallel & & \shortparallel \\
 & \omega_{2j+1,\,0} & \omega_{2j+1,\,2} & \cdots & \omega_{2j+1,\,2J-2j-2}
\end{array}
    - The final output is: $\boldsymbol{w}=\big(w_1, w_2, \dots, w_J\big)'=\big(\omega_{1,\,0}, \omega_{3,\,0}, \dots, \omega_{2J-1,\,0}\big).$
- Every iteration applies only scalar multiplication by $2$, subtraction, and array indexing — all linear operations over $\mathbb{R}^J$, which is consistent with the fact that $\mathcal{T}_J^{(A)}$ is a linear-transformation. Consequently, $\mathcal{T}_J^{(A)}$ satisfies the relation $\mathcal{T}\big(\alpha\,\boldsymbol{u}+\beta\,\boldsymbol{v}\big)=\alpha\,\mathcal{T}(\boldsymbol{u})+\beta\,\mathcal{T}(\boldsymbol{v})$ for all $\boldsymbol{u},\boldsymbol{v}\in\mathbb{R}^J$ and $\alpha,\beta\in\mathbb{R}$.

```python
def type_A_transformation(u: ndarray) -> ndarray:
    """
    Implementing the main functional component of type A transformation involving Chebyshev's polynomials to a 1D array.
    """
    
    # Initialize the output array
    T: ndarray = u.copy()
    
    # Get the number of terms in the array
    J: int = T.shape[0]
    
    # Initialize an array to store and track the terms w_{2j, 1} for j = 0, ..., J-1
    E: ndarray = T.copy()
    
    # Initialize an array to store and track the terms w_{2j+1, 0} for j = 0, ..., J-1
    O: ndarray = T.copy()
   
    # Iterate over each integer j = 0, ..., J-1
    for j in range(J):
        
        # Store the (j+1)-th output element: w_{2j+1, 0}
        T[j] = O[0]
        
        # Update terms in E using the recurrence: w_{2j, 2k-1} = 2*w_{2j-1, 2k} - w_{2j-2, 2k-1}
        E[:J-j-1] = 2.0 * O[1:J-j] - E[:J-j-1]
            
        # Update terms in O using the recurrence: w_{2j+1, 2k-2} = 2*w_{2j, 2k-1} - w_{2j-1, 2k-2}
        O[:J-j-1] = 2.0 * E[:J-j-1] - O[:J-j-1]

    return T
```

Since the Chebyshev polynomials are iteratively computed from a recurrence relation of degree $2$, we require two iterator variables `E` and `O` to keep track of the transformation. The variable `E` tracks even first index terms like $\omega_{2j,\,2k-1}$ which appear in the first step of each iteration, and the variable `O` tracks odd first index terms like $\omega_{2j-1,\,2k}$ which appear in the second step of each iteration.

### Example Usage

The following examples illustrate two canonical applications of the transformation to an odd power basis, and a general linearity verification.

**Cosine test.** We evaluate the transformation on the odd cosine-power vector
$\big(\cos(x),\,\cos^3(x),\,\cos^5(x),\,\cos^7(x),\,\cos^9(x)\big)'$
for a randomly drawn $x\sim\mathrm{Uniform}(0,1)$, and verify that the output matches
$\big(\cos(x),\,\cos(3x),\,\cos(5x),\,\cos(7x),\,\cos(9x)\big)'$,
consistent with the identity $T_{2j-1}\big(\cos(x)\big)=\cos\big((2j-1)\,x\big)$.


```python
x = random.uniform()
powers = array([cos(x)**(2*j-1) for j in range(1, 6)])
result = type_A_transformation(powers)
expected = array([cos((2*j-1)*x) for j in range(1, 6)])

print("Output:   ", result)
print("Expected: ", expected)
```

**Sine test.** We also evaluate the transformation on the odd sine-power vector
$\big(\sin(x),\,\sin^3(x),\,\sin^5(x),\,\sin^7(x),\,\sin^9(x)\big)'$
for the same value of $x$, and verify that the output matches
$\big(\sin(x),\,-\sin(3x),\,\sin(5x),\,-\sin(7x),\,\sin(9x)\big)'$,
consistent with the identity $T_{2j-1}\big(\sin(x)\big)=(-1)^{j-1}\,\sin\big((2j-1)\,x\big)$.


```python
powers = array([sin(x)**(2*j-1) for j in range(1, 6)])
result = type_A_transformation(powers)
expected = array([((-1)**(j-1)) * sin((2*j-1)*x) for j in range(1, 6)])

print("Output:   ", result)
print("Expected: ", expected)
```

**Linear-transformation test:** Since `type_A_transformation` implements a linear map $\mathcal{T}_J^{(A)}:\mathbb{R}^J\to\mathbb{R}^J$, it accepts any vector $\boldsymbol{u}\in\mathbb{R}^J$ as input — not just odd power basis vectors. We verify the identity $\mathcal{T}_J^{(A)}\big(\alpha\,\boldsymbol{u}+\beta\,\boldsymbol{v}\big)=\alpha\,\mathcal{T}_J^{(A)}(\boldsymbol{u})+\beta\,\mathcal{T}_J^{(A)}(\boldsymbol{v})$ on two arbitrary vectors $\boldsymbol{u},\boldsymbol{v}$ and scalars $\alpha,\beta$.

```python
tau = type_A_transformation
u = random.normal(size=5)
v = random.normal(size=5)
alpha = random.normal()
beta = random.normal()

print("𝜏(𝛼u+𝛽v):   ", tau(alpha*u+beta*v))
print("𝛼𝜏(u)+𝛽𝜏(v):", alpha*tau(u)+beta*tau(v))
```

## Type B Transformation

This function implements type B transformation involving Chebyshev's polynomials to a 1D array, as discussed in _Chakravorty, A., et al. (2021)_.

### Parameters

- **ndarray**  
  `u`: 1D input array $\boldsymbol{u}=\big(u_1, u_2,\dots, u_J\big)'\in\mathbb{R}^J$.

### Returns

- **ndarray**  
  `T`: The transformed 1D array $\mathcal{T}_J^{(B)}(\boldsymbol{u})\in\mathbb{R}^J$, the index $J$ in $\mathcal{T}_J^{(B)}$ means it expects 1D input array of $J$ elements.

### Notes

- The function implements a linear-transformation $\mathcal{T}_J^{(B)}:\mathbb{R}^J\to\mathbb{R}^J$, equivalent to left-multiplication by a $J\times J$ invertible matrix $\mathsf{B}_J$, which can be expressed in terms of the coefficients from Chebyshev's orthogonal polynomials. This is formally verified in _Chakravorty, A., et al. (2021)_.
- When the input is the power basis vector $\boldsymbol{u}\,(t)=\big(1,\,t,\dots,t^{J-1}\big)'$, the output is the Chebyshev polynomial basis $\mathcal{T}_J^{(B)}\big(\boldsymbol{u}\,(t)\big)=\big(T_0(t),T_1(t),\dots,T_{J-1}(t)\big)'$, where $T_{n}(t)$ is the $n$th Chebyshev polynomial of the first kind. Also: $T_{j}\big(\cos(x)\big)=\cos(j\,x)$ for $x\in\mathbb{R}$ and $j\in\mathbb{Z}_{\geq 0}$.
- For the canonical input $\boldsymbol{u}\,(t)=\big(1,t^1,\dots,t^{J-1}\big)'$, observe that $T_0(t)=1$ and $T_1(t)=t$. Consequently, we start the transformation with two vectors: $\big(T_0(t),t\,T_0(t),\dots,t^{J-1}\,T_0(t)\big)'$ and $\big(T_1(t),t\,T_1(t),\dots,t^{J-2}\,T_1(t)\big)'$. The transformation consists of $J$ iterative steps as follows:
    - We start the $j$th iteration with the $(J-j+1)$-dimensional vector: $\big(t^0\,T_{j-1}(t),t^1\,T_{j-1}(t),\dots,t^{J-j}\,T_{j-1}(t)\big)'$ and the $(J-j)$-dimensional vector: $\big(t^0\,T_j(t),t^1\,T_j(t),\dots,t^{J-j-1}\,T_j(t)\big)'$, we add $T_{j-1}(t)$ as the $j$th element of the final-output vector.
    - Then, we use the recurrence relation: $t^{k-1}\,T_{j+1}(t)=2\,t^k\,T_j(t)-t^{k-1}\,T_{j-1}(t)$, to calculate $T_{j+1}(t),t\,T_{j+1}(t),\dots,t^{J-j-2}\,T_{j+1}(t)$ from $T_{j-1}(t),t\,T_{j-1}(t),\dots,t^{J-j-2}\,T_{j-1}(t)$ and $t\,T_j(t),t^2\,T_j(t),\dots,t^{J-j-1}\,T_j(t)$. The following diagram demonstrates this process:
\begin{array}{ccccc}
 & -1\,\times\,T_{j-1}(t) & -1\,\times\,t\,T_{j-1}(t) & \cdots & -1\,\times\,t^{J-j-2}\,T_{j-1}(t) \\
 & + & + & & + \\
 & \quad 2\,\times\,t\,T_j(t) & \quad 2\,\times\,t^2\,T_j(t) & \cdots & \quad 2\,\times\,t^{J-j-1}\,T_j(t) \\
 & \shortparallel & \shortparallel & & \shortparallel \\
 & T_{j+1}(t) & t\,T_{j+1}(t) & \cdots & t^{J-j-2}\,T_{j+1}(t)
\end{array}
- For a general input $\boldsymbol{u}=\big(u_1,u_2,\dots,u_J\big)'$, we set $\omega_{0,\,k-1}=u_k$ for $1\leq k\leq J$ and $\omega_{1,\,k-1}=u_k$ for $1\leq k< J$. We begin the transformation with $\big(\omega_{0,\,0},\omega_{0,\,1},\dots,\omega_{0,\,J-1}\big)'$ and $\big(\omega_{1,\,0},\omega_{1,\,1},\dots,\omega_{1,\,J-2}\big)'$ and perform $J$ iterative steps, mimicking the logic for canonical input as follows:
    - We start the $j$th iteration with the $(J-j+1)$-dimensional vector $\big(\omega_{j-1,\,0},\omega_{j-1,\,1},\dots,\omega_{j-1,\,J-j}\big)'$ and the $(J-j)$-dimensional vector $\big(\omega_{j,\,0},\omega_{j,\,1},\dots,\omega_{j,\,J-j-1}\big)'$, and we add $w_j=\omega_{j-1,\,0}$ as the $j$th element of the final-output vector.
    - Then, use the recurrence relation: $\omega_{j+1,\,k-1}=2\,\omega_{j,\,k}-\omega_{j-1,\,k-1}$, to calculate $\omega_{j+1,\,0},\omega_{j+1,\,1},\dots,\omega_{j+1,\,J-j-2}$ from $\omega_{j-1,\,0},\omega_{j-1,\,1},\dots,\omega_{j-1,\,J-j-2}$ and $\omega_{j,\,1},\omega_{j,\,2},\dots,\omega_{j,\,J-j-1}$. The following diagram demonstrates this process:
\begin{array}{ccccc}
 & -1\times\omega_{j-1,\,0} & -1\times\omega_{j-1,\,1} & \cdots & -1\times\omega_{j-1,\,J-j-2} \\
 & + & + & & + \\
 & \quad 2\times\omega_{j,\,1} & \quad 2\times\omega_{j,\,2} & \cdots & \quad 2\times\omega_{j,\,J-j-1} \\
 & \shortparallel & \shortparallel & & \shortparallel \\
 & \omega_{j+1,\,0} & \omega_{j+1,\,1} & \cdots & \omega_{j+1,\,J-j-2}
\end{array}
    - The final output is: $\boldsymbol{w}=\big(w_1,w_2,\dots,w_J\big)'=\big(\omega_{0,\,0},\omega_{1,\,0},\dots,\omega_{J-1,\,0}\big).$
- Every iteration applies only scalar multiplication by $2$, subtraction, and array indexing — all linear operations over $\mathbb{R}^J$, which is consistent with the fact that $\mathcal{T}_J^{(B)}$ is a linear-transformation. Consequently, $\mathcal{T}\big(\alpha\,\boldsymbol{u}+\beta\,\boldsymbol{v}\big)=\alpha\,\mathcal{T}(\boldsymbol{u})+\beta\,\mathcal{T}(\boldsymbol{v})$ for all $\boldsymbol{u},\boldsymbol{v}\in\mathbb{R}^J$ and $\alpha,\beta\in\mathbb{R}$.
- We use iterator variables `E` and `O` to track the changes in the for loops.

```python
def type_B_transformation(u: ndarray) -> ndarray:
    """
    Apply type B transformation involving Chebyshev's polynomials to a 1D array.
    """
    
    # Initialize the output array
    T: ndarray = u.copy()
    
    # Get the number of terms in the array
    J: int = T.shape[0]
    
    # Initialize an array to store and track the terms w_{2j, 0} for j = 0, ..., floor(J/2)-1
    E: ndarray = T.copy()
    
    # Initialize an array to store and track the terms w_{2j+1, 0} for j = 0, ..., floor(J/2)-1
    O: ndarray = T[1:].copy()
    
    # Iterate over each integer j = 0, ..., floor(J/2) - 1
    for j in range(J // 2):
        
        # Store the first element of E as the (2j)-th output element: w_{2j, 0}
        T[2*j]   = E[0]
        
        # Store the first element of O as the (2j+1)-th output element: w_{2j+1, 0}
        T[2*j+1] = O[0]
        
        # Update terms in E using the recurrence: w_{2j, 2k-1} = 2*w_{2j-1, 2k} - w_{2j-2, 2k-1}
        E[:J-2*j-2] = 2.0 * O[1:J-2*j-1] - E[:J-2*j-2]
        
        # Update terms in O using the recurrence: w_{2j+1, 2k-2} = 2*w_{2j, 2k-1} - w_{2j-1, 2k-2}
        if J-2*j-3 > 0:
            O[:J-2*j-3] = 2.0 * E[1:J-2*j-2] - O[:J-2*j-3]
    
    # If J is odd, store the first element of E as the (J-1)-th output element: w_{J-1, 0}
    if J % 2 == 1:
        T[J-1] = E[0]

    return T
```

Even though each iteration involves only one step, we require two iterator variables `E` and `O` to keep track of the transformation, since the Chebyshev polynomials are iteratively computed from a recurrence relation of degree $2$. The variable `E` tracks even first index terms like $\omega_{2j,\,k}$ which appear in even iterations, and the variable `O` tracks odd first index terms like $\omega_{2j-1,\,k}$ which appear in odd iterations, and there is a final assignment if $J$ is odd.

### Example Usage

The following examples illustrate a canonical application of the transformation to a consecutive power basis, and a general linearity verification.

**Cosine test.** We evaluate the transformation on the consecutive cosine-power vector
$\big(1,\,\cos(x),\,\cos^2(x),\,\cos^3(x),\,\cos^4(x)\big)'$
for a randomly drawn $x\sim\mathrm{Uniform}(0,1)$, and verify that the output matches
$\big(1,\,\cos(x),\,\cos(2x),\,\cos(3x),\,\cos(4x)\big)'$,
consistent with the identity $T_{j}\big(\cos(x)\big)=\cos(jx)$.


```python
x = random.uniform()
powers = array([cos(x)**j for j in range(5)])
result = type_B_transformation(powers)
expected = array([cos(j*x) for j in range(5)])

print("Output:   ", result)
print("Expected: ", expected)
```

**Linear-transformation test:** Since `type_B_transformation` implements a linear map $\mathcal{T}_J^{(B)}:\mathbb{R}^J\to\mathbb{R}^J$, it accepts any vector $\boldsymbol{u}\in\mathbb{R}^J$ as input — not just consecutive power basis vectors. We verify the identity $\mathcal{T}_J^{(B)}\big(\alpha\,\boldsymbol{u}+\beta\,\boldsymbol{v}\big)=\alpha\,\mathcal{T}_J^{(B)}(\boldsymbol{u})+\beta\,\mathcal{T}_J^{(B)}(\boldsymbol{v})$ on two arbitrary vectors $\boldsymbol{u},\boldsymbol{v}$ and scalars $\alpha,\beta$.

```python
tau = type_B_transformation
u = random.normal(size=5)
v = random.normal(size=5)
alpha = random.normal()
beta = random.normal()

print("𝜏(𝛼u+𝛽v):   ", tau(alpha*u+beta*v))
print("𝛼𝜏(u)+𝛽𝜏(v):", alpha*tau(u)+beta*tau(v))
```

## Sine Power Transformation

This function implements a sine power transformation to a 1D array, as discussed in _Chakravorty, A., et al. (2021)_.

### Parameters

- **ndarray**  
  `u`: 1D input array $\boldsymbol{u}=\big(u_1, u_2,\dots, u_J\big)'\in\mathbb{R}^J$.

### Returns

- **ndarray**  
  `T`: The transformed 1D array $\mathcal{T}_J^{(S)}(\boldsymbol{u})\in\mathbb{R}^J$, the index $J$ in $\mathcal{T}_J^{(S)}$ means it expects 1D input array of $J$ elements.

### Notes

- The function implements a linear-transformation equivalent to left-multiplication by a $J\times J$ lower-triangular matrix with ones on the diagonal (hence always invertible). This is formally verified in _Chakravorty, A., et al. (2021)_.
- When the input is the mixed monomial vector $\boldsymbol{u}\,(x)=\big(\sin(x),\;\sin(x)\cos^2(x),\;\dots,\;\sin(x)\cos^{2J-2}(x)\big)'$, the output is the odd sine power vector $\mathcal{T}_J^{(S)}\big(\boldsymbol{u}\,(x)\big)=\big(\sin(x),\,\sin^3(x),\,\dots,\,\sin^{2J-1}(x)\big)'$.
- We recursively use the identity $\sin^{2j+1}(x)\,\cos^{2k-2}(x)=\sin^{2j-1}(x)\,\cos^{2k-2}(x)-\sin^{2j-1}(x)\,\cos^{2k}(x)$ for $j\in\mathbb{Z}_{> 0}$ and $k\in\mathbb{Z}_{> 0}$, in $J-1$ iterative steps to carry out the transformation. In the $j$th iteration, we calculate the $(J-j)$-dimensional vector $\big(\sin^{2j+1}(x),\,\sin^{2j+1}(x)\,\cos^2(x),\,\dots,\,\sin^{2j+1}(x)\,\cos^{2J-2j-2}(x)\big)$ as the difference of the $(J-j)$-dimensional vectors $\big(\sin^{2j-1}(x),\,\sin^{2j-1}(x)\,\cos^2(x),\,\dots,\,\sin^{2j-1}(x)\,\cos^{2J-2j-2}(x)\big)$ and $\big(\sin^{2j-1}(x)\,\cos^2(x),\,\sin^{2j-1}(x)\,\cos^4(x),\,\dots,\,\sin^{2j-1}(x)\,\cos^{2J-2j}(x)\big)$.

- For a general input $\boldsymbol{u}=\big(u_1, u_2, \dots, u_J\big)'$, we mimic the logic of the transformation for canonical input. We set $w_1=u_1$, $u_k = \omega_{1,\,2k-2}$ for $1\leq k\leq J$ and use the recurrence-relation: $\omega_{2j+1,\,2k-2}=\omega_{2j-1,\,2k-2}-\omega_{2j-1,\,2k}$, in $J-1$ iterative steps to carry out the transformation. In the $j$th iteration, we calculate the $(J-j)$-dimensional vector $\big(\omega_{2j+1,0},\,\omega_{2j+1,2},\,\dots,\,\omega_{2j+1,2J-2j-2}\big)$ as the difference of the $(J-j)$-dimensional vectors $\big(\omega_{2j-1,0},\,\omega_{2j-1,2},\,\dots,\,\omega_{2j-1,2J-2j-2}\big)$ and $\big(\omega_{2j-1,2},\,\omega_{2j-1,4},\,\dots,\,\omega_{2j-1,2J-2j}\big)$, and set $w_{j+1}=\omega_{2j+1,0}$. The final output is $\boldsymbol{w}=\big(w_1, w_2, \dots, w_J\big)'$.
- Every iteration applies only subtraction and array indexing — all linear operations over $\mathbb{R}^J$, which is consistent with the fact that the transformation is linear. Consequently, $\mathcal{T}\big(\alpha\,\boldsymbol{u}+\beta\,\boldsymbol{v}\big)=\alpha\,\mathcal{T}(\boldsymbol{u})+\beta\,\mathcal{T}(\boldsymbol{v})$ for all $\boldsymbol{u},\boldsymbol{v}\in\mathbb{R}^J$ and $\alpha,\beta\in\mathbb{R}$.

```python
def sine_transformation(u: ndarray) -> ndarray:
    """
    Implement sine power transformation to a 1D array.
    """

    # Initialize the output array
    T: ndarray = u.copy()
    
    # Get the number of terms in the array
    J: int = T.shape[0]
  
    # Iterate over each integer k = 1, ..., J-1
    for k in range(1, J):
        
        # Update using the recurrence: T[j] = T[j-1] - T[j]
        T[k:] = T[k-1:J-1] - T[k:]

    return T
```

## Alternating Sign Transformation

This function applies an alternating sign transformation to a 1D array.

### Parameters

- **ndarray**  
  `u`: 1D input array $\boldsymbol{u}=\big(u_1, u_2,\dots, u_J\big)'\in\mathbb{R}^J$.

### Returns

- **ndarray**  
  `T`: The transformed 1D array with $u_j$ replaced by $(-1)^{j-1}\cdot u_j$ for $j=1,\dots,J$.

### Notes

- The transformation is diagonal and therefore trivially linear and $\mathcal{T}\big(\alpha\,\boldsymbol{u}+\beta\,\boldsymbol{v}\big)=\alpha\,\mathcal{T}(\boldsymbol{u})+\beta\,\mathcal{T}(\boldsymbol{v})$ for all $\boldsymbol{u},\boldsymbol{v}\in\mathbb{R}^J$ and $\alpha,\beta\in\mathbb{R}$.

```python
def alterneate_sign_transformation(u: ndarray) -> ndarray:
    """
    Implement alternating sign transformation to a 1D array.
    """
    
    # Initialize the output array
    T: ndarray = u.copy()
    
    # Get the number of terms in the array
    J: int = T.shape[0]
    
    # Iterate over each integer j = 0, ..., J-1
    for j in range(J):
        # Update using the recurrence: T[j] = (-1)^j * T[j]
        T[j] = (-1)**j * T[j]
    
    return T
```

### Example Usage

The following examples illustrate the full sine-branch pipeline composition and a general linearity verification.

**Full sine pipeline.** The three sine-branch transformations compose as follows:

$$
\big(\sin(x),\,\sin(x)\cos^2(x),\,\sin(x)\cos^4(x),\,\dots\big)'
\xrightarrow{\texttt{sine\_transformation.}}
\big(\sin(x),\,\sin^3(x),\,\sin^5(x),\,\dots\big)'
\xrightarrow{\texttt{type\_A\_transformation}}
\big(\sin(x),\,-\sin(3x),\,\sin(5x),\,\dots\big)'
\xrightarrow{\texttt{alternate\_sign\_transformation}}
\big(\sin(x),\,\sin(3x),\,\sin(5x),\,\dots\big)'
$$


```python
def _sine_pipeline_transformation_(u: ndarray) -> ndarray:
    """
    Implement the entire transformation pipeline for sine
    """

    # Apply Sine Power Transformation to the input
    T = sine_transformation(u)

    # Apply Type A Transformation to the result
    T = type_A_transformation(T)

    # Apply Alternating Sign Transformation to the result
    T = alterneate_sign_transformation(T)
    
    return T
```

to $\big(\sin(x),\,\sin(3x),\,\sin(5x),\,\sin(7x),\,\sin(9x)\big)'$ for a randomly drawn $x$. Because all three constituent maps are linear and invertible, their composition is also a linear, invertible map, and the input is uniquely recoverable from the output.

We verify the identity $\mathcal{T}_J^{(A)'}\big(\alpha\,\boldsymbol{u}+\beta\,\boldsymbol{v}\big)=\alpha\,\mathcal{T}_J^{(A)'}(\boldsymbol{u})+\beta\,\mathcal{T}_J^{(A)'}(\boldsymbol{v})$ on two arbitrary vectors $\boldsymbol{u},\boldsymbol{v}$ and scalars $\alpha,\beta$.

## Remark

To avoid any confusion, we would like to point out that the linear-transformation $\tau_J^{(A)}$ in _Lemma G.1._ from the supplementary-material of _Chakravorty, A., et al. (2021)_, is not exactly equivallent to the transformation $\mathcal{T}_J^{(A)}$, implemented in the method `type_A_transformation`. From the statement of _Lemma G.1._, observe that, $\tau_J^{(A)}:\mathbb{R}^{2J}\to\mathbb{R}^{2J}$ is a linear transformation which takes the canonical-input: $\big(\cos(x), \sin(x), \cos^3(x), \sin(x)\,\cos^2(x),\dots,\cos^{2J-1}(x), \sin(x)\,\cos^{2J-2}(x)\big)'$, and transforms it to the canonical output: $\big(\cos(x), \sin(x), \cos(3x), \sin(3x),\dots,\cos\big((2J-1)\,x\big), \sin\big((2J-1)\,x\big)\big)'$. We can implement the transformation $\tau_J^{(A)}$ in a function `_type_A_transformation_full_`, using the functions `type_A_transformation` and `_sine_pipeline_transformation_` as follows

```python
def _type_A_transformation_full_(u):

    # Extract odd members of the input
    u_odd = u[range(0, len(u), 2)]

    # Apply Type A Transformation to the result
    T_odd = type_A_transformation(u_odd)

    # Extract even members of the input
    u_even = u[range(1, len(u), 2)]

    # Apply Sine Pipeline Transformation to the result
    T_even = _sine_pipeline_transformation_(u_even)

    # Stack odd and even transformed values in the final output
    T = vstack([T_odd, T_even]).T.ravel()
    
    return T
```

**Canonical example test:**
We verify that the function `_type_A_transformation_full_` transforms the canonical-input: $\big(\cos(x), \sin(x), \cos^3(x), \sin(x)\,\cos^2(x),\dots,\cos^{2J-1}(x), \sin(x)\,\cos^{2J-2}(x)\big)'$ to the canonical output: $\big(\cos(x), \sin(x), \cos(3x), \sin(3x),\dots,\cos\big((2J-1)\,x\big), \sin\big((2J-1)\,x\big)\big)'$, for arbitrary $x$

```python
x = random.uniform()
powers = array([[cos(x)**(2*j+1), sin(x)*cos(x)**(2*j)] for j in range(5)]).ravel()
result = _type_A_transformation_full_(powers)
expected = array([[cos((2*j+1)*x), sin((2*j+1)*x)] for j in range(5)]).ravel()

print("Output:  ", result)
print("Expected:", expected)
```

**Linear-transformation test:** Finally, since `_type_A_transformation_full_` implements the linear transformation, say $\tau_J^{(A)}:\mathbb{R}^{2J}\to\mathbb{R}^{2J}$, it takes any vector $\boldsymbol{u}\in\mathbb{R}^{2J}$ as input. We verify the identity $\tau_J^{(A)}\big(\alpha\,\boldsymbol{u}+\beta\,\boldsymbol{v}\big)=\alpha\,\tau_J^{(A)'}(\boldsymbol{u})+\beta\,\tau_J^{(A)'}(\boldsymbol{v})$ on two arbitrary vectors $\boldsymbol{u},\boldsymbol{v}$ and scalars $\alpha,\beta$

```python
tau = _type_A_transformation_full_
u = random.normal(size=10)
v = random.normal(size=10)
alpha = random.normal()
beta = random.normal()

print("𝜏(𝛼u+𝛽v):   ", tau(alpha*u+beta*v))
print("𝛼𝜏(u)+𝛽𝜏(v):", alpha*tau(u)+beta*tau(v))
```

## Cosine-Product-Tensor to Cosine-Vector Transformation

This function is designed to transform a $K$-dimensional tensor of cosine products into a 1D vector of cosines by recursively applying the cosine product-to-sum formula (Type C transformation), as described in _Chakravorty, A., et al. (2021)_. We denote this transformation as $\mathcal{T}_{\boldsymbol{J}}^{(\cos)}(\,.)$, where $\boldsymbol{J}=(J_1,J_2,\dots,J_K)'$ is the $K$ dimensional integer-parameter vector.

### Parameters

- **ndarray**  
  `U`: $K$-dimensional input tensor $\boldsymbol{U}\in\mathbb{R}^{J_1\times J_2\times\dots\times J_K}$. For a general index-vector $\boldsymbol{j}=(j_1,j_2,\dots,j_K)'$, where $j_k\in\{1,2,\dots,J_k\}$ for $1\leq k\leq K$, the $\boldsymbol{j}$th element of $\boldsymbol{U}$ is $u_{j_1,\,j_2,\,\dots,\,j_K}$.

### Returns

- **ndarray**  
  `T`: 1D vector $\mathcal{T}_{\boldsymbol{J}}^{(\cos)}(\boldsymbol{U})\in\mathbb{R}^J$ of length $J=\prod_{k=1}^K J_k$.

### Notes

- The function implements a linear-transformation $\mathcal{T}_{\boldsymbol{J}}^{(\cos)}:\mathbb{R}^{J_1\times J_2\times\dots\times J_K}\to\mathbb{R}^J$, where $J = \prod_{k=1}^K J_k$. The input-tensor shape $\boldsymbol{J}:=(J_1, J_2, \dots, J_K)$ can be any arbitrary positive integer-vector with length $K > 1$. This transformation is formally described in _Chakravorty, A., et al. (2021)_, as a component of _Type C_ transformation.
- The transformation $\mathcal{T}_{\boldsymbol{J}}^{(\cos)}$ is generic; the shape-vector $\boldsymbol{J}$ is not explicitly part of the input, but is inferred from the input tensor.
- For the canonical example, when the $\boldsymbol{j}$th entry is $u_{j_1,\,j_2,\,\dots,\,j_K}(x)=\cos\big((2j_{1}-1)\,x\big)\,\prod_{k=2}^K\cos(j_k\,L_k\,x)$, with $L_k=2\,\prod_{k=1}^K J_k$ for $k=2,\dots,K$; the output is a 1D vector of length $J=\prod_{k=1}^K J_k$, whose $j$th element is $\cos\big((2j-1)\,x\big)$. Hence, the function implements a linear transformation from the $K$-dimensional cosine product tensor to a 1D cosine vector. The cosine product basis and the 1D odd cosine basis span the same function space under the given frequency scaling, so the map is invertible.
- For a general input tensor, the function collapses the last dimension iteratively using the product-to-sum identity: $\cos(A+B)=2\,\cos(A)\,\cos(B)-\cos(A-B)$. The general idea is, at each step, the $j$th slice is combined with the running accumulator via an expression similar to: `2 * transformed_slice[j] - accumulator[::-1]`, where the reversal implements the $\cos(A-B)$ term.
- **Base case ($K=2$):** 
    - We start the transformation with an empty list of 1D vectors: $T$, assign the first column of the input matrix: $\boldsymbol{U}_{:\,1}$ to the accumulator variable: $t$, add $t$ to the list $T$, and perform $J_2-1$ iterative steps as follows:
        -  At $j$th iteration we update the accumulator as: $t=2*\boldsymbol{U}_{:\,j}-t^R$, here $\boldsymbol{U}_{:\,j}$ is the $j$th column of the input matrix, and, $t^R$ denote the vector obtained by reversing the elements of $t$.
        -  Add $t$ to the list $T$.
    - At the end, combine all 1D vectors in $T$ in one single 1D vector, and return that vector as function output.
- **Recursive case ($K>2$):**
    - When $K>2$, we recursively define $\mathcal{T}_{\boldsymbol{J}}^{(\cos)}(\,.)$ in terms of $\mathcal{T}_{\boldsymbol{J}_{1:(K-1)}}^{(\cos)}(\,.)$ where $\boldsymbol{J}_{1:(K-1)}$ is the $(K-1)$-dimensional integer vector obtained by deleting the last element of $\boldsymbol{J}$.
    - We begin with an empty list of 1D vectors: $T$, assign the transformed first slice along last dimension of the input tensor: $\mathcal{T}_{\boldsymbol{J}_{1:(K-1)}}^{(\cos)}\big(\boldsymbol{U}_{::\,1}\big)$ to the accumulator variable: $t$, and perform $J_2-1$ iterative steps as follows:
        -  At $j$th iteration we update the accumulator as: $t=2*\mathcal{T}_{\boldsymbol{J}_{1:(K-1)}}^{(\cos)}\big(\boldsymbol{U}_{::\,j}\big)-t^R$, here $\boldsymbol{U}_{:\,j}$ is the $j$th slice along last dimension of the input tensor.
        -  Add $t$ to the list $T$.
    - At the end, combine all 1D vectors in $T$ in one single 1D vector, and return that vector as function output.
- Every step applies only scalar multiplication by $2$, subtraction, and array reversal — all linear operations, consistent with the fact that the full map is a linear transformation. Consequently, $\mathcal{T}\big(\alpha\,\boldsymbol{U}+\beta\,\boldsymbol{V}\big)=\alpha\,\mathcal{T}(\boldsymbol{U})+\beta\,\mathcal{T}(\boldsymbol{V})$ for all input tensors $\boldsymbol{U},\boldsymbol{V}$ and $\alpha,\beta\in\mathbb{R}$.  

```python
def transform_cosine_product_tensor_to_cosine_vector(U: ndarray) -> ndarray:
    """
    Transform a K-dimensional tensor of cosine products to a 1D vector of cosine terms.
    """
    # Import necessary Numpy function
    from numpy import array
    
    # Check if input is 2D (base case)
    if len(U.shape) == 2:
        # Initialize result list with size equal to last dimension
        T = []
        
        # Extract first column as initial accumulator
        t = U[:, 0]
        
        # Store first column in result
        T.append(t)
        
        # Iterate over each column j = 1, ..., J_K - 1
        for j in range(1, U.shape[-1]):
            # Update using the recurrence: 2*cos(A)cos(B) - reflected accumulator
            t = 2 * U[:, j] - t[::-1]
            
            # Store updated accumulator in result
            T.append(t)
        
        # Flatten result list to 1D array
        T = array(T).ravel()
    
    # Recursive case for K > 2 dimensions
    else:
        # Initialize result list with size equal to the last dimension
        T = []
        
        # Recursively transform first slice along last dimension
        t = transform_cosine_product_tensor_to_cosine_vector(U[..., 0])
        
        # Store first transformed slice in result
        T.append(t)
        
        # Iterate over each slice j = 1, ..., J_K - 1 along last dimension
        for j in range(1, U.shape[-1]):
            # Recursively transform slice and update using the recurrence
            t = 2 * transform_cosine_product_tensor_to_cosine_vector(U[..., j]) - t[::-1]
            
            # Store updated accumulator in result
            T.append(t)
        
        # Flatten result list to 1D array
        T = array(T).ravel()
    
    return T
```

**Canonical example test.** We construct a 3-dimensional input tensor of shape $\boldsymbol{J}:=(J_1,\,J_2,\,J_3)'=(5,\,10,\,10)'$
whose $\boldsymbol{j}:=(j_1,\,j_2,\,j_3)'$th entry is

- Every step applies only scalar multiplication by $2$, subtraction, and array reversal — all linear operations, consistent with the fact that the full map is a linear transformation. Consequently, $\mathcal{T}\big(\alpha\,\boldsymbol{U}+\beta\,\boldsymbol{V}\big)=\alpha\,\mathcal{T}(\boldsymbol{U})+\beta\,\mathcal{T}(\boldsymbol{V})$ for all input tensors $\boldsymbol{U},\boldsymbol{V}$ and $\alpha,\beta\in\mathbb{R}$.  

```python
def transform_sine_product_tensor_to_sine_vector(U: ndarray) -> ndarray:
    """
    Transform a K-dimensional tensor of sine products to a 1D vector of sine terms.
    """
    # Import necessary Numpy function
    from numpy import array
    
    # Check if input is 2D (base case)
    if len(U.shape) == 2:
        # Initialize result list with size equal to last dimension
        T = []
        
        # Extract first column as initial accumulator
        t = U[:, 0]
        
        # Store first column in result
        T.append(t)
        
        # Iterate over each column j = 1, ..., J_K - 1
        for j in range(1, U.shape[-1]):
            # Update using the recurrence: 2*sin(A)cos(B) - reflected accumulator
            t = 2 * U[:, j] + t[::-1]
            
            # Store updated accumulator in result
            T.append(t)
        
        # Flatten result list to 1D array
        T = array(T).ravel()
    
    # Recursive case for K > 2 dimensions
    else:
        # Initialize result list with size equal to the last dimension
        T = []
        
        # Recursively transform first slice along last dimension
        t = transform_sine_product_tensor_to_sine_vector(U[..., 0])
        
        # Store first transformed slice in result
        T.append(t)
        
        # Iterate over each slice j = 1, ..., J_K - 1 along last dimension
        for j in range(1, U.shape[-1]):
            # Recursively transform slice and update using the recurrence
            t = 2 * transform_sine_product_tensor_to_sine_vector(U[..., j]) + t[::-1]
            
            # Store updated accumulator in result
            T.append(t)
        
        # Flatten result list to 1D array
        T = array(T).ravel()
    
    return T
```

## General Trigonometric-Power-Tensor  to Trigonometric-Vector transformation

- This function converts a $K$-dimensional tensor of trigonometric powers into a 1D vector of trigonometric terms. 
- It implements the linear-transformation $\tau_{\boldsymbol{J}}$ as described in _Theorem 5.1_ of _Chakravorty, A., et al. (2021)_.

### Parameters
- `U` **: numpy.ndarray**  
  $K$-dimensional input tensor $\boldsymbol{U}$ of shape: $\big(2\,J_{1}, J_{2}, \cdots, J_{K}\big)$.
### Returns

- **numpy.ndarray**  
  `T`: 1D vector $\tau_{\boldsymbol{J}}(\boldsymbol{U})\in\mathbb{R}^J$ of length $2\,J=2\,\prod_{k=1}^K J_k$.

### Notes

- The function implements the linear-transformation $\tau_{\boldsymbol{J}}:\mathbb{R}^{2\,J_1\times J_2\times\dots\times J_K}\to\mathbb{R}^{2\,J}$, where $J = \prod_{k=1}^K J_k$. The input-tensor shape $\boldsymbol{J}:=\big(2\,J_1, J_2, \dots, J_K\big)$ can be any arbitrary positive integer-vector with length $K > 1$. This transformation from _Theorem 5.1_ of _Chakravorty, A., et al. (2021)_ is the fundamental linear transformation for speeding up serial SEP computation using Chebyshev's orthogonal polynomials.
- The transformation $\tau_{\boldsymbol{J}}$ is generic; the shape-vector $\boldsymbol{J}$ is not explicitly part of the input, but is inferred from the input tensor.
- For the canonical case, when the $\big(2\,j_1-1,j_2,\dots,j_k\big)'$th and the $\big(2\,j_1,j_2,\dots,j_k\big)'$th entries of the input tensor $U$ are $\cos^{2j_1-1}(x)\,\prod_{k=2}^K\cos^{j_k-1}(L_k\,x)$ and $\cos(x)\,\sin^{2j_1-2}(x)\,\prod_{k=2}^K\cos^{j_k-1}(L_k\,x)$ respectively, where $L_k=2\,\prod_{k=1}^K J_k$ for $k=2,\dots,K$; the output is a 1D vector of length $2\,\prod_{k=1}^K J_k$, whose $(2j-1)$th and $(2j)$th elements are $\cos\big((2j-1)\,x\big)$ and $\sin\big((2j-1)\,x\big)$ respectively. 
- The function first splits the input tensor $U$ into two $K$-dimensional sub-tensors $U^{\cos}$ and $U^{\sin}$ corresponding to even and odd indices along the first index of $U$. Observe that both $U^{\cos}$ and $U^{\sin}$ has shape $\big(J_1, J_2, \dots, J_K\big)$. Then each of these tensors is transformed separately as follows
   - **Cosine tensor:**
       - The linear transformation: `type_A_transformation` is applied along the axis $1$ of $U^{\cos}$.
       - The linear transformation: `type_B_transformation` is sequentially applied along the axes $2,\dots, K$ of $U^{\cos}$.
       - The linear transformation: `transform_cosine_product_tensor_to_cosine_vector` is applied to $U^{\cos}$.
       - Observe that each of these linear transformations changes the contents of $U^{\cos}$; the final version of $U^{\cos}$ is a vector of size $\prod_{k=1}^K J_k$.
   - **Sine tensor:**
       - The linear transformation: `sine_transformation` is applied along the axis $1$ of $U^{\sin}$.
       - The linear transformation: `type_A_transformation` is applied along the axis $1$ of $U^{\sin}$.
       - The linear transformation: `alterneate_sign_transformation` is applied along the axis $1$ of $U^{\sin}$.
       - The linear transformation: `type_B_transformation` is sequentially applied along the axes $2,\dots, K$ of $U^{\sin}$.
       - The linear transformation: `transform_cosine_product_tensor_to_cosine_vector` is applied to $U^{\sin}$.
       - Again, observe that each of these linear transformations changes the contents of $U^{\sin}$; the final version of $U^{\sin}$ is a vector of size $\prod_{k=1}^K J_k$.
    - Finally, $U^{\cos}$ and $U^{\sin}$ are stacked together in a single vector $T$ of size $2\,\prod_{k=1}^K J_k$ so that the even and the odd indices correspond to the cosine and sine terms.
    - $T$ is returned as output.

```python
def transform_trigonometric_power_tensor_to_trigonometric_vector(U):
    """
    Converts a K-dimensional tensor of trigonometric powers into a 1D vector of trigonometric terms.
    """
    # Import all the necessary Python modules
    import numpy
    
    # Import "epstatKDTree" module
    import epstatKDTree
    
    # Get the dimensions of the input tensor
    J = U.shape
    
    # Cosine powers: even indices along first axis
    U_cos = U[range(0, J[0], 2), ...]
    
    # Apply type A transformation along the first axis
    U_cos = numpy.apply_along_axis(epstatKDTree.type_A_transformation, 0, U_cos)
    
    # Apply type B transformation along the remaining axes
    for d in range(1, len(J)):
        U_cos = numpy.apply_along_axis(epstatKDTree.type_B_transformation, d, U_cos)
    
    # Transform product of cosine SEP to cosine SEP
    U_cos = epstatKDTree.transform_cosine_product_tensor_to_cosine_vector(U_cos)
    
    # Sine SEP: odd indices along first axis
    U_sin = U[range(1, J[0], 2), ...]
    
    # Apply sine transformation along the first axis
    U_sin = numpy.apply_along_axis(epstatKDTree.sine_transformation, 0, U_sin)
    
    # Apply type A transformation along the first axis
    U_sin = numpy.apply_along_axis(epstatKDTree.type_A_transformation, 0, U_sin)
    
    # Change signs of odd entries along the first axis
    U_sin = numpy.apply_along_axis(epstatKDTree.alterneate_sign_transformation, 0, U_sin)
    
    # Apply type B transformation along the remaining axes
    for d in range(1, len(J)):
        U_sin = numpy.apply_along_axis(epstatKDTree.type_B_transformation, d, U_sin)
    
    # Transform product of sine SEP to sine SEP
    U_sin = epstatKDTree.transform_sine_product_tensor_to_sine_vector(U_sin)
    
    # Collect all (cosine, sine) pairs of trigonometric SEP for each index in a 1D array and concatenate the multiplier as the first element
    T = numpy.stack((U_cos, U_sin), axis = 1).ravel()
    return(T)
```

```python
def transform_1D_trigonometric_product_vector_to_trigonometric_vector(U, J):
    """
    Converts a K-dimensional tensor of trigonometric product SEP statistics into a 1D vector of trigonometric SEP statistics.
    """
    # Import all the necessary Python modules
    import numpy
    
    # Import "epstatKDTree" module
    import epstatKDTree
    
    # First dimension doubles due to cos/sin split
    J_ = numpy.array(J)
    J_[0] = 2 * J[0]
    
    # Fetch the first term
    u = U[0]
    
    # Reshape the input into a tensor for transformations
    U = numpy.delete(U, 0).reshape(J_)

    # Transform the tensor U using "epstatKDTree.transform_trigonometric_power_tensor_to_trigonometric_vector()"
    T = epstatKDTree.transform_trigonometric_power_tensor_to_trigonometric_vector(U)
    
    # Collect all (cosine, sine) pairs of trigonometric SEP for each index in a 1D array and concatenate the multiplier as the first element
    T = numpy.insert(T, 0, u)
    
    return(T)
```

## Multi-Axis Trigonometric SEP Transformation for $k$-d tree

- This function transforms a $P$-dimensional tensor of product of trigonometric powers into a $P$-dimensional tensor of product of trigonometric terms by applying the `transform_1D_trigonometric_product_vector_to_trigonometric_vector` function along each axis iteratively. 
- It performs the linear-transformation as described in _Lemma 1.1_ of the supplemetary-text of _Chakravorty, A., et al. (2022)_.

### Parameters
- `U` **: numpy.ndarray**  
  A $P$-dimensional tensor of shape $\big(\,2\, \prod_{k=1}^K J_k + 1\,\big)^P$, where each axis encodes flattened trigonometric product SEP statistics.
- `J` **: array_like of int**  
  A sequence (tuple, list, or 1mD array) of length $K$ specifying the dimensions $(J_1, J_2, \dots, J_K)$ of the original $K$-dimensional trigonometric expansion.

### Returns
- `T` **: **numpy.ndarray**  
  A tensor of the same shape $\big(\,2\, \prod_{k=1}^K J_k + 1\,\big)^P$, where along each axis the coefficients represent trigonometric SEP statistics instead of trigonometric product SEP statistics.

### Notes
- The dimensionality $P$ is inferred from the shape of `U` using: `int(numpy.log(U.shape[0]) / log(2 * numpy.prod(J) + 1))`.
- The function `transform_1D_trigonometric_product_vector_to_trigonometric_vector` is applied sequentially along each axis, converting trigonometric powers into standard trigonometric terms.

```python
def transform_PD_trigonometric_product_tensor_to_trigonometric_tensor(U, J):
    """
    Transforms a P-dimensional tensor of trigonometric product SEP statistics into trigonometric SEP statistics
    by applying `transform_1D_trigonometric_product_vector_to_trigonometric_vector` along each axis.
    """
    # Import all the necessary Python modules
    import numpy
    
    # Import "epstatKDTree" module
    import epstatKDTree

    # Infer P = tensor dimensionality based on total length reduction ratio
    P = int(numpy.log(U.shape[0]) / numpy.log(2 * numpy.prod(J) + 1))

    # Wrapper for axis-wise transformation
    def lambda_warpper(U_axis):
        return epstatKDTree.transform_1D_trigonometric_product_vector_to_trigonometric_vector(U_axis, J)

    # Apply transformation sequentially along all axes
    for d in range(P):
        U = numpy.apply_along_axis(lambda_warpper, d, U)
    return U
```
